```python
import pandas as pd
df = pd.read_csv('/workspaces/data_3500_sandbox/lecture_code/titanic.csv')
df
```

```python

"""
Programming Activity 1

Using the Titanic dataset and the Pandas library, fill in any missing data cells in the 'ages' column with the median age.
"""

df["Age"] = df["Age"].fillna(df["Age"].median())
df
```

```python
"""
Programming Activity 2

Using the Titanic dataset and the Pandas library, create a new column called 'FamilySize' and fill it with the size of the passenger's family.
"""

df["FamilySize"] = df["Name"].notna().astype(int) + df["SibSp"]
df
```

```python
"""
Programming Activity 3

Using the Titanic dataset and the Pandas library, find the average survival rate by class and sex.
"""

survival_rate = df.groupby(["Pclass", "Sex"])["Survived"].mean()
print("Avg survival rate by class and sex: ", survival_rate)
```

```python
"""
Programming Activity 4

Using the Titanic dataset and the Pandas library, find the average fare per class.
"""

Avg_Fare_Per_Class = df.groupby(["Pclass"])["Fare"].mean()
print(Avg_Fare_Per_Class)
```

```python
"""
Programming Activity 5

Using the Titanic dataset and the Pandas library, find the median age of survivors versus non-survivors.
"""
df["Survived"] = df["Survived"].map({0: "Did Not Survive", 1: "Survived"})

Median_Age = df.groupby(["Survived"])["Age"].median()
print(Median_Age)
```

```python
df2 = pd.read_csv('/workspaces/data_3500_sandbox/lecture_code/SLC_weather.csv')
df2
```

```python
"""
Programming Activity 6
Using the weather file (located on week 11 programming activities page), find the longest warm streak where the high temperature was above 65 degrees.

"""
Temp_Streak = df2["Max Temp"]

df2["Warm"] = df2["Max Temp"] > 65
df2["Streak"] = (df2["Warm"] != df2["Warm"].shift()).cumsum()

Streaks = df2[df2["Warm"]].groupby("Streak").size()

Long_Streak = Streaks.max()
print("The longest streak of days over 65 degrees was", Long_Streak, "days.")
```

```python
"""
Programming Activity 7

Using the same weather dataset, find the hottest day and the coldest night.
"""

Warmest_IDX = df2["Max Temp"].idxmax()
Coldest_IDX = df2["Min Temp"].idxmin()

Warmest_Day = df2.loc[Warmest_IDX]
Coldest_Day = df2.loc[Coldest_IDX]

print("The warmest day of the year was: ", Warmest_Day)
print("The coldest day of the year was: ", Coldest_Day)

```

```python
"""
Programming Activity 8
Using the same weather dataset, find the single most volatile day (the day with the biggest difference between high and low temperature).

"""

Temp_Diff = (df2["Max Temp"] - df2["Min Temp"]).max()
Temp_Diff_Date = df2.loc[(df2["Max Temp"] - df2["Min Temp"]).idxmax(), "Date"]

print("The largest temperature difference in one day was", Temp_Diff, "degrees, and it was on", Temp_Diff_Date)
```
